# 1. 클래스와 객체

클래스는 쉽게 말하면 **객체를 만들기 위한 설계도**이다.

```python
class Student:
    pass
```

여기서 `Student`는 실제 학생이 아니라
**학생 객체를 만들기 위한 설계도**이다.

실제 객체는 다음과 같이 만든다.

```python
student1 = Student()
student2 = Student()
```

구조를 간단하게 표현하면 다음과 같다.

```text
Student 클래스
     ↓
   설계도
   ↙   ↘
student1 student2
  객체     객체
```

같은 `Student` 클래스로 만들었더라도
`student1`과 `student2`는 서로 다른 객체이다.

# 3. 같은 클래스로 만든 객체는 같은 객체일까?

다음 코드를 실행했다.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1 is student2)
```

결과:

```text
False
```

처음에는 같은 `Student` 클래스로 만들었는데
왜 `False`가 나오는지 헷갈렸다.

하지만 `Student()`를 실행할 때마다 **새로운 객체가 생성**된다.

```text
student1 → Student 객체 A

student2 → Student 객체 B
```

`is`는 두 객체의 내용이 비슷한지를 확인하는 것이 아니라,

> 두 변수가 정말 동일한 객체를 가리키고 있는가?

를 확인한다.

따라서:

```python
student1 is student2
```

는 `False`이다.

반대로:

```python
student1 = Student()
student2 = student1

print(student1 is student2)
```

라면 두 변수가 같은 객체를 가리키므로:

```text
True
```

가 된다.

# 4. `type()`은 무엇을 확인할까?

다음 코드도 실행했다.

```python
class Student:
    pass

student = Student()

print(student)
print(type(student))
```

`print(student)`의 결과는 대략 다음처럼 나온다.

```text
<__main__.Student object at 0x...>
```

처음에는 오류처럼 보였지만 오류가 아니다.

파이썬이 `student`가 가리키고 있는 **객체 자체의 정보**를 보여주는 것이다.

반면:

```python
print(type(student))
```

의 결과는:

```text
<class '__main__.Student'>
```

처럼 나온다.

`type()`은 쉽게 말해서:

> 이 값은 어떤 종류인가?

를 확인하는 함수이다.

예를 들어:

```python
number = 10
name = "홍길동"

print(type(number))
print(type(name))
```

결과:

```text
<class 'int'>
<class 'str'>
```

그리고:

```python
student = Student()
print(type(student))
```

라고 하면 `student`는 `Student` 클래스로 만든 객체이기 때문에
타입에 `Student`가 표시된다.

### 기억하기

```text
type(10)
→ int

type("홍길동")
→ str

type(student)
→ Student
```

# 5. `__init__()`으로 객체의 처음 값 저장하기

학생에게 이름과 점수 같은 데이터를 넣어보기 시작했다.

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
```

그리고 객체를 만든다.

```python
student1 = Student("홍길동", 99)
```

이렇게 하면:

```text
"홍길동" → name
99       → score
```

가 전달된다.

결과적으로 `student1` 객체에는 다음 데이터가 저장된다.

```text
student1
 ├── name  → "홍길동"
 └── score → 99
```

# 6. `self`란?

오늘 가장 생소했던 개념 중 하나가 `self`였다.

```python
self.name
self.score
```

`self`는 쉽게 생각하면:

> 현재 이 객체

를 의미한다.

예를 들어:

```python
student1 = Student("홍길동", 99)
```

라는 객체가 있을 때 `student1`의 메서드를 실행하면
그 메서드 안의 `self`는 현재 `student1`을 가리킨다.

```text
self.name
↓
현재 객체의 name

self.score
↓
현재 객체의 score
```

따라서:

```python
student1.name
```

은:

```text
홍길동
```

이고:

```python
student1.score
```

는:

```text
99
```

이다.

# 7. 속성(Attribute)

객체가 가지고 있는 데이터를 **속성(attribute)**이라고 한다.

예를 들어:

```python
class Student:
    def __init__(self, name, score, age):
        self.name = name
        self.score = score
        self.age = age
```

객체를 만든다.

```python
student1 = Student("홍길동", 99, 34)
```

그러면:

```text
student1
 ├── name  → 홍길동
 ├── score → 99
 └── age   → 34
```

가 된다.

속성은 `.`을 이용해서 접근한다.

```python
student1.name
student1.score
student1.age
```

# 9. 메서드(Method)

클래스 안에 정의된 함수를 **메서드(method)**라고 한다.

다음과 같이 학생의 결과를 출력하는 기능을 만들었다.

```python
def show_result(self):
    print(f"{self.name}: {self.score}점")
```

객체에서 메서드를 실행하려면:

```python
student1.show_result()
```

라고 작성한다.

결과:

```text
홍길동: 99점
```

쉽게 읽으면:

```text
student1.show_result()

student1아!
네 결과를 보여줘!
```

라고 생각할 수 있다.

# 12. `return` 이후의 코드는 실행되지 않는다

다음과 같이 작성하면서 문제가 발생했다.

```python
def is_passed(self):
    return self.score >= 60
    print("합격입니다")
```

`return`은 결과를 돌려주는 동시에
**함수를 종료한다.**

따라서:

```text
return 실행
     ↓
함수 종료
     ↓
아래 print() 실행 안 됨
```

이 된다.

조건에 따라 메시지를 출력하고 싶다면 `if`를 먼저 사용할 수 있다.

```python
def is_passed(self):
    if self.score >= 60:
        print(f"{self.name}님 합격입니다. 축하합니다!")
```

# 16. `students`와 `student`의 차이

오늘 실제로 다음 오류가 발생했다.

```text
AttributeError: 'list' object has no attribute 'age'
```

문제가 된 형태는:

```python
students.age
```

였다.

하지만 `students`는 학생 한 명이 아니다.

```python
students = [student1, student2]
```

즉 **리스트 전체**이다.

```text
students → 학생 여러 명이 들어 있는 리스트 📦

student → 리스트에서 현재 꺼낸 학생 한 명 👤
```

그래서:

```python
students.age
```

는 사용할 수 없다.

반복문 안에서는:

```python
student.age
student.name
student.score
```

를 사용해야 한다.

### 핵심

```text
student1 → 홍길동 객체 한 명

student2 → 강감찬 객체 한 명

students → 두 객체를 담고 있는 리스트

student → for문에서 현재 꺼낸 객체 한 명
```

# 18. 오늘 완성한 형태

오늘 배운 내용을 종합하면 다음과 같은 구조를 만들 수 있었다.

```python
class Student:
    def __init__(self, name, score, age):
        self.name = name
        self.score = score
        self.age = age

    def show_result(self):
        print(f"{self.name}: {self.score}점")

    def is_passed(self):
        return self.score >= 60


student1 = Student("홍길동", 99, 34)
student2 = Student("강감찬", 58, 29)

students = [student1, student2]

for student in students:

    if student.is_passed():
        result = "당당히 합격"
    else:
        result = "아쉽지만 탈락"

    print(f"{student.age}세 {student.name}님 {result}")
    print(
        f"{student.age}세 {student.name}님의 점수는 "
        f"{student.score}점 입니다"
    )
```

예상 결과:

```text
34세 홍길동님 당당히 합격
34세 홍길동님의 점수는 99점 입니다

29세 강감찬님 아쉽지만 탈락
29세 강감찬님의 점수는 58점 입니다
```
