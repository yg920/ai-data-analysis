# 2. 지출 데이터의 구조

```python
expense = {
    "date": "2026-09-01",
    "category": "식비",
    "description": "점심",
    "amount": 12000
}
```

여기서 지출 한 건은 **딕셔너리**로 표현했다.

# 3. 여러 지출은 리스트에 저장

지출은 한 건만 존재하는 것이 아니기 때문에 여러 딕셔너리를 리스트에 저장한다.

```python
expenses = []
```

지출을 추가하면

```python
expenses.append(expense)
```

처럼 리스트에 새로운 딕셔너리가 들어간다.

구조를 그림으로 보면 다음과 같다.

```text
expenses
  │
  ├─ 지출 1
  │   ├─ date
  │   ├─ category
  │   ├─ description
  │   └─ amount
  │
  ├─ 지출 2
  │   ├─ date
  │   ├─ category
  │   ├─ description
  │   └─ amount
  │
  └─ 지출 3
```

즉,

> **지출 한 건 = 딕셔너리**
> **여러 지출 = 리스트**

라고 이해할 수 있다.

# 7. 숫자 입력 예외 처리

금액은 숫자로 입력받아야 한다.

```python
amount = int(input("금액: "))
```

하지만 사용자가

```text
만원
abc
오천원
```

처럼 입력하면 `int()`로 변환할 수 없다.

이때 발생하는 오류가

```text
ValueError
```

이다.

그래서 다음과 같이 처리했다.

```python
try:
    amount = int(input("금액: "))
except ValueError:
    print("금액은 숫자만 입력!")
    return
```

# 8. try-except 복습

구조는 다음과 같다.

```python
try:
    오류가 발생할 수도 있는 코드

except 오류이름:
    오류가 발생했을 때 실행할 코드
```

오늘 프로그램에서는

```python
try:
    amount = int(input("금액: "))
except ValueError:
```

를 사용했다.

즉,

```text
정상 숫자 입력
    ↓
int() 변환 성공
    ↓
계속 진행
```

잘못된 입력:

```text
abc
 ↓
int() 변환 실패
 ↓
ValueError 발생
 ↓
except 실행
```

# 9. 값의 범위 검증

숫자로 변환됐다고 해서 모든 숫자가 올바른 것은 아니다.

예:

```text
0
-5000
```

지출 금액으로 사용할 수 없도록 다음 조건을 사용했다.

```python
if amount <= 0:
    print("금액은 0보다 큰 값으로!")
    return
```

여기서 중요한 차이가 있다.

```text
"abc" 입력
→ 숫자로 변환할 수 없음
→ ValueError

-1000 입력
→ 숫자로 변환 가능
→ ValueError 아님
→ 조건문으로 검사
```

즉,

> **예외 처리와 값 검증은 서로 다른 역할을 한다.**
