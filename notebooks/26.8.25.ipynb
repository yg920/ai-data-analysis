# 1. 빈 집합과 빈 딕셔너리

처음 헷갈렸던 부분이다.

```python
empty_set = set()   # 빈 집합
empty_dict = {}     # 빈 딕셔너리
```

`{}`만 작성하면 빈 집합이 아니라 **빈 딕셔너리**가 만들어진다.

빈 집합을 만들려면 반드시 `set()`을 사용해야 한다.

```python
print(type(empty_set))
print(type(empty_dict))
```

정리하면 다음과 같다.

```text
set() → 빈 집합
{}    → 빈 딕셔너리
```

# 2. 딕셔너리란?

딕셔너리는 데이터를 **키(Key) : 값(Value)** 형태로 저장하는 자료형이다.

```python
student = {
    "name": "민수",
    "score": 90
}
```

여기에서:

```text
"name"  → 키(Key)
"민수"  → 값(Value)

"score" → 키(Key)
90      → 값(Value)
```

즉, 키를 이용해서 원하는 값을 찾을 수 있다.

---

# 3. 리스트와 딕셔너리의 차이

리스트에서는 **인덱스 번호**를 이용해서 데이터를 찾는다.

```python
student = ["민수", 90]

print(student[0])
```

결과:

```text
민수
```

반면 딕셔너리는 **키**를 이용한다.

```python
student = {
    "name": "민수",
    "score": 90
}

print(student["name"])
```

결과:

```text
민수
```

따라서 간단하게 기억하면:

```text
리스트     → 위치(인덱스)로 찾는다.
딕셔너리   → 키(Key)로 찾는다.
```

# 7. `[]`, `get()`, `in` 구분하기

오늘 배운 내용 중 특히 중요했던 부분이다.

## 반드시 존재해야 하는 키 → `[]`

```python
print(student["name"])
```

키가 존재한다면 값을 가져온다.

하지만 해당 키가 없다면:

```text
KeyError
```

가 발생한다.

---

## 없을 수도 있는 키 → `get()`

```python
print(student.get("city"))
```

`city`라는 키가 없어도 오류가 발생하지 않고 기본적으로:

```text
None
```

을 반환한다.

기본값을 직접 정할 수도 있다.

```python
print(student.get("city", "정보 없음"))
```

키가 없다면:

```text
정보 없음
```

이 반환된다.

---

## 키가 존재하는지 먼저 확인 → `in`

```python
if "score" in student:
    print(student["score"])
```

`in`은 딕셔너리에 해당 **키가 존재하는지** 확인할 때 사용할 수 있다.

### 선택 기준

```text
반드시 필요한 키       → []
없을 수도 있는 키      → get()
키 존재 여부 먼저 확인 → in
```

```python
if "city" in student:
```

같은 방법을 사용할 수 있다.

# 10. for문과 딕셔너리

`items()`와 `for`문을 함께 사용하면 키와 값을 하나씩 꺼낼 수 있다.

```python
dic = {
    "파이썬": 4,
    "AI": 2
}

for word, count in dic.items():
    print(word, count)
```

반복되는 과정을 생각하면:

```text
첫 번째 반복
word = "파이썬"
count = 4

두 번째 반복
word = "AI"
count = 2
```

가 된다.

여기서 중요한 점은:

```text
word  → 키
count → 값
```

이라는 것이다.

# 19. += 1의 의미

처음 작성할 때 다음과 같이 헷갈렸다.

```python
dic[word] =+ 1
```

하지만 기존 값에 `1`을 더하려면:

```python
dic[word] += 1
```

이라고 작성해야 한다.

즉:

```python
dic[word] += 1
```

은 쉽게 말하면:

> `dic[word]`의 현재 값에 `1`을 더해서 다시 저장한다.

라는 의미다.

예를 들어:

```text
2 → 3 → 4 → 5
```

처럼 값이 증가한다.

# 20. 완성된 빈도수 계산 부분

지금까지의 내용을 합치면:

```python
for word in words:
    if word not in dic:
        dic[word] = 1
    else:
        dic[word] += 1
```

이 된다.

이 코드를 한국어로 설명하면:

> `words`에서 단어를 하나씩 꺼내 `word`에 넣는다.
> `word`가 `dic`의 키에 없다면 `word`를 새로운 키로 저장하고 값을 `1`로 만든다.
> 이미 `word`가 `dic`의 키에 있다면 해당 키의 값에 `1`을 더한다.

# 24. 오늘 과제 최종 코드

```python
sentense = "파이썬 어렵다 파이썬 힘들다 파이썬 즐겁다 파이썬 그렇다"

words = sentense.split()

dic = {}

for word in words:
    if word not in dic:
        dic[word] = 1
    else:
        dic[word] += 1

print(dic)

for word, count in dic.items():
    if count >= 3:
        print(f"3번 이상 사용된 단어는 {word}이고, 반복된 횟수는 {count}번입니다.")
```

# 25. 테스트하면서 발견한 중요한 점

처음 테스트 문장으로 다음과 같은 문장을 생각했다.

```text
파이썬에 파이썬에의한 파이썬을 위한 정의
```

사람이 보면 모두 `파이썬`이라는 단어가 포함되어 있다.

하지만 현재 프로그램에서는 `split()`을 사용하고 있기 때문에:

```text
파이썬에
파이썬에의한
파이썬을
```

을 각각 **서로 다른 문자열**로 판단한다.

즉:

```text
"파이썬에" != "파이썬에의한"
"파이썬에의한" != "파이썬을"
```

이다.

컴퓨터는 사람이 문장의 의미를 이해하는 것처럼 판단하는 것이 아니라 **현재 코드에서 정한 기준에 따라 정확한 문자열을 비교한다.**

# 28. 오늘 문제를 해결한 사고 과정

# 29. 오늘 느낀 문제 해결의 핵심

오늘 과제를 하면서 느낀 것은 **문법을 아는 것과 문제를 해결하는 것은 조금 다르다**는 것이다.

나는 이미:

```python
for
if
else
in
딕셔너리
items()
f-string
```

등을 배웠다.

하지만 처음 문제를 봤을 때 이것들을 어떤 순서로 연결해야 하는지가 바로 떠오르지는 않았다.

그래서 문제를 코드로 바로 바꾸려고 하기보다 먼저:

```text
무엇을 가지고 있는가?
↓
무엇을 구해야 하는가?
↓
중간에 어떤 정보가 필요한가?
↓
그 정보를 어떤 자료형에 저장할 것인가?
↓
어떤 조건으로 판단할 것인가?
↓
결과가 맞는지 어떻게 검증할 것인가?
```

순서로 생각하는 것이 도움이 되었다.

# 30. AI를 공부에 활용하는 방법

오늘은 AI에게 처음부터 완성 코드를 요청하지 않았다.

대신:

```text
문제 확인
↓
한 단계 힌트
↓
직접 코드 작성
↓
틀린 부분 확인
↓
다시 힌트
↓
다시 작성
↓
결과 예측
↓
코드의 의미를 직접 설명
```

방식으로 문제를 해결했다.

이렇게 공부하면 단순히 코드를 복사하는 것이 아니라 **왜 이 코드가 필요한지를 생각하는 연습**을 할 수 있다.

특히 AI에게:

> "정답을 알려줘."

라고 하기보다:

> "현재 내가 생각한 방향이 맞는지 확인해줘."

> "다음 단계 힌트 하나만 줘."

> "완성 코드는 알려주지 말고 틀린 부분만 찾아줘."

처럼 요청하는 것이 학습에 더 도움이 될 수 있다는 것을 알게 되었다.

# 31. 오늘의 핵심 정리

```text
{}                       → 빈 딕셔너리

딕셔너리[키] = 값         → 값 저장

word in dic              → word라는 키가 있는지 확인

word not in dic          → word라는 키가 없는지 확인

dic[word] = 1            → 처음 등장한 단어를 1로 저장

dic[word] += 1           → 기존 등장 횟수에 1 추가

dic.keys()               → 키

dic.values()             → 값

dic.items()              → 키와 값

for word in words        → 단어를 하나씩 확인

for word, count in dic.items()
                         → 키와 값을 하나씩 확인

if count >= 3            → 등장 횟수가 3 이상인지 확인

get()                    → 없는 키도 안전하게 조회

[]                       → 반드시 존재해야 하는 키 조회
```
