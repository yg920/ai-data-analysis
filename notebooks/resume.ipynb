# 1. 관심 분야

현재 내가 가장 관심 있는 분야는 **AI·AX 서비스 기획 및 운영, AI 기반 프로젝트 관리, AI Product 직무**이다.

기존에는 공공기관에서 SNS 및 뉴미디어 채널을 운영하면서 콘텐츠 기획, 서비스 운영, 이용자 커뮤니케이션, 데이터 기반 성과 분석 등의 업무를 경험했다.

현재는 KANT AX Agent Engineer 과정을 통해 Python, SQL, PostgreSQL, Django, 데이터 분석, AI Agent, LLM 활용, 서비스 개발 등을 학습하고 있다.

따라서 기존의 운영·기획 경험을 완전히 버리고 순수 개발자로 전환하기보다는,

**서비스 운영 경험 + 데이터 활용 능력 + AI 기술 이해 + 프로젝트 기획 능력**

을 결합할 수 있는 직무가 가장 적합하다고 판단하였다.

현재 관심 직무는 다음과 같다.

1. AI/AX 서비스 운영
2. AI Project Manager
3. AI Product Owner / Product Manager
4. AI 기반 B2B 서비스 운영 및 컨설팅
5. 장기적으로 AI Product Manager / Technical PM

나의 목표는 단순히 AI 관련 회사에 취업하는 것이 아니라 **AI 기술을 실제 사용자의 문제 해결이나 기업의 업무 자동화에 연결하는 역할**을 담당하는 것이다.

**준비할 것**

KANT 프로젝트를 AI 서비스 운영 관점에서도 설명할 수 있도록 정리하고, 기존 SNS 운영 경력을 단순 콘텐츠 제작이 아니라 **사용자 운영·서비스 성장 경험**으로 재구성할 예정이다.

### 포지션 판단

현재 경험을 가장 직접적으로 활용하면서 더 좋은 AI 기업으로 이동할 수 있는 포지션이다.

단순히 SNS 업무를 계속하는 것이 아니라 **AI 서비스 운영 → Product/AX 기획**으로 이동할 수 있는 발판이 될 수 있다고 생각한다.

공공기관 사업 및 콘텐츠 운영 경험이 있어 사용자 관점에서 서비스와 업무를 분석해 본 경험도 있다.

Python, SQL, DB, Django 등 개발 기초를 학습하고 있다.

**부족한 부분**

엔터프라이즈 AI 플랫폼에 대한 기술적 이해가 아직 부족하다.

클라우드, Docker, Kubernetes, MLOps 등의 이해가 부족하다.

제품 개발 프로젝트를 실제 고객 환경에 적용한 경험도 없다.

### 준비할 것

Docker 및 클라우드 기본 학습을 진행한다.

REST API 및 외부 서비스 연동 프로젝트를 추가한다.

AI Agent 프로젝트를 하나 완성해 배포하고 기술 구조를 설명할 수 있도록 준비한다.

### 포지션 판단

지금 당장 지원하기보다는 **AI/AX 관련 회사에서 1~2년 정도 프로젝트 및 제품 경험을 쌓은 뒤 도전하고 싶은 직무**이다.

장기적으로는 단순 서비스 운영보다 이런 Technical PM 수준까지 성장하고 싶다.

# 6. 주관적 준비사항

내가 현재 가장 보완해야 한다고 생각하는 것은 **프로젝트를 설명하는 능력**이다.

현재 여러 프로젝트를 진행했지만 면접관에게

“왜 이 프로젝트를 했는가?”

“어떤 문제를 해결했는가?”

“내가 어떤 역할을 했는가?”

“기술적으로 어떤 선택을 했는가?”

“결과적으로 어떤 효과가 있었는가?”

를 짧고 명확하게 설명하는 연습이 아직 부족하다.

또한 새로운 AI·AX 직무에서는 기존 SNS 경력이 전혀 다른 경력으로 보일 수도 있기 때문에 이를 연결하는 스토리가 필요하다.

따라서 기존 업무를 단순히

**SNS 콘텐츠를 운영했다**

가 아니라

**사용자 데이터를 분석하고 여러 채널의 서비스를 운영하면서 문제를 발견하고 개선해 성과를 만들어냈다**

라는 방식으로 표현할 필요가 있다.

앞으로는 프로젝트 하나를 3분 이내로 설명하는 연습과 포트폴리오 발표 연습을 진행할 예정이다.

Product와 AX 직무에서도 데이터 기반 의사결정이 반복적으로 요구되었다.

SQL을 이용해 사용자 행동과 KPI를 분석하는 프로젝트를 하나 추가한다.

## 3위. AI 서비스 기획 문서 작성 경험

현재 진행 중인 프로젝트를 기준으로 PRD, 요구사항 정의서, 사용자 Flow, 기능 우선순위 등을 작성한다.

## 5위. 포트폴리오와 프로젝트 발표 능력 강화

프로젝트마다 3분 설명 자료를 만든다.

문제 → 역할 → 기술 → 결과 → 배운 점 순서로 설명하는 연습을 진행한다.

# 11. 이번 조사를 하면서 느낀 점

이번 조사를 하기 전에는 AI 분야로 취업하려면 Python이나 머신러닝을 매우 잘해야 하고 개발자가 되어야 한다고 생각한 부분이 있었다.

하지만 실제 채용공고를 확인해 보니 AI 산업에는 개발자 외에도 다양한 역할이 있었다.

특히 AI 기술을 이해하면서 고객의 문제를 찾고, 서비스를 운영하고, 데이터를 분석하고, 개발자와 협업하여 실제 제품을 만들어가는 역할이 많았다.

이 부분에서 내가 기존에 경험한 공공기관 SNS와 서비스 운영 업무도 완전히 버려야 할 경력이 아니라는 것을 알게 되었다.

오히려 기존의

**서비스 운영 + 콘텐츠 기획 + 커뮤니케이션 경험**

에 현재 배우고 있는

**Python + SQL + DB + AI Agent + 데이터 분석**

을 결합하면 나만의 경쟁력을 만들 수 있다고 생각한다.

따라서 당장의 목표를 순수 AI 개발자로 한정하지 않고,

**AI 서비스 운영 → AX 프로젝트/서비스 기획 → AI Product Owner 또는 Technical PM**

으로 이어지는 경력 경로를 만들어 보고 싶다.

KANT 과정을 수료할 때까지 단순히 수업을 많이 듣는 것보다 실제로 보여줄 수 있는 AI Agent 프로젝트와 포트폴리오를 완성하는 것이 가장 중요하다고 생각한다.

이번 채용공고 조사를 통해 앞으로 무엇을 공부해야 할지가 이전보다 구체적으로 보이기 시작했다.
