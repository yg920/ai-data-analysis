# 1. Git이란?

Git은 프로젝트 파일의 **변경 이력을 관리하는 버전 관리 시스템**이다.

예를 들어 Python 프로젝트를 만들면서 코드를 수정했을 때 Git을 사용하면

```text
처음 작성한 코드
↓
기능 추가
↓
오류 수정
↓
기능 개선
```

처럼 변경된 기록을 관리할 수 있다.

Git을 사용하면 잘못 수정했을 때 이전 상태를 확인하거나 여러 사람이 함께 개발할 때 서로의 변경 내용을 관리하기 쉬워진다.

---

# 2. Git과 GitHub의 차이

처음에는 Git과 GitHub가 같은 것처럼 느껴질 수 있지만 역할이 다르다.

| 구분         | 역할                            |
| ---------- | ----------------------------- |
| Git        | 내 컴퓨터에서 코드 변경 이력 관리           |
| GitHub     | Git으로 관리하는 프로젝트를 인터넷에 저장하고 공유 |
| Repository | 프로젝트 저장 공간                    |
| Commit     | 변경 내용을 하나의 기록으로 저장            |
| Push       | 내 컴퓨터의 Git 기록을 GitHub로 전송     |
| Pull       | GitHub의 변경 내용을 내 컴퓨터로 가져오기    |

쉽게 표현하면

```text
내 컴퓨터
   ↓
  Git
   ↓
GitHub Repository
```

라고 이해할 수 있다.

라는 의미다.

# 4. Git 상태 확인

현재 Git 상태는 다음 명령어로 확인한다.

```bash
git status
```

오늘 확인했던 결과는 다음과 같았다.

```text
On branch main
nothing to commit, working tree clean
```

의미는 다음과 같다.

### `On branch main`

현재 사용하고 있는 브랜치가 `main`이라는 의미이다.

### `nothing to commit`

새롭게 Commit할 변경 내용이 없다는 의미이다.

### `working tree clean`

현재 Git이 추적하고 있는 파일과 실제 프로젝트 파일의 상태가 동일하다는 의미이다.

즉,

```text
오류가 발생한 상태 ❌

변경된 파일이 없는 정상 상태 ⭕
```

이다.

# 5. GitHub Remote Repository

내 컴퓨터의 Git 저장소와 GitHub Repository를 연결하기 위해서는 Remote 주소를 등록한다.

예:

```bash
git remote add origin https://github.com/사용자명/저장소명.git
```

현재 연결되어 있는 Remote 주소는 다음 명령어로 확인할 수 있다.

```bash
git remote -v
```

# 6. 오늘 발생한 GitHub Remote 오류

오늘 Push 과정에서 다음 오류가 발생했다.

```text
remote: Not Found

fatal: repository
'https://github.com/yg920/https://github.com/yg920/python_basic/'
not found
```

주소를 자세히 보면 문제가 보인다.

```text
https://github.com/yg920/
https://github.com/yg920/python_basic/
```

GitHub 주소가 **두 번 붙어 있었다.**

잘못된 Remote URL 때문에 GitHub에서 Repository를 찾지 못한 것이다.

---

# 7. Remote URL 수정

먼저 현재 Remote 주소를 확인한다.

```bash
git remote -v
```

그리고 잘못된 주소를 수정한다.

```bash
git remote set-url origin https://github.com/yg920/python_basic.git
```

다시 확인한다.

```bash
git remote -v
```

정상적으로 다음과 비슷하게 나오면 된다.

```text
origin  https://github.com/yg920/python_basic.git (fetch)
origin  https://github.com/yg920/python_basic.git (push)
```

이후 GitHub에 올린다.

```bash
git push -u origin main
```

# 8. Git Push 흐름

오늘 배운 GitHub 업로드 과정을 하나의 흐름으로 정리하면 다음과 같다.

```text
프로젝트 생성
     ↓
git init
     ↓
git add .
     ↓
git commit
     ↓
GitHub Repository 생성
     ↓
git remote add origin
     ↓
git push
```

대표적인 명령어:

```bash
git init
git add .
git commit -m "first commit"
git remote add origin 저장소주소
git push -u origin main
```

# 10. Markdown이란?

Markdown은 문서를 간단한 문법으로 작성하는 방식이다.

GitHub의

```text
README.md
```

파일도 Markdown으로 작성한다.

대표적인 문법은 다음과 같다.

---

## 제목

```markdown
# 제목 1

## 제목 2

### 제목 3
```

---

## 글자 강조

```markdown
**굵은 글씨**

*기울임*
```

결과:

**굵은 글씨**

*기울임*

---

## 목록

```markdown
- Python
- PostgreSQL
- Git
- GitHub
```

---

## 번호 목록

```markdown
1. Python
2. PostgreSQL
3. Git
```

---

## 코드

한 줄 코드는 백틱을 사용한다.

```markdown
`print("Hello")`
```

여러 줄 코드는 백틱 3개를 사용한다.

````markdown
```python
print("Hello")
print("Python")
```
````

# 11. GitHub Markdown Preview

GitHub에서 Markdown 형식으로 작성했더라도 파일이 Markdown 파일로 인식되어야 Preview가 정상적으로 표시된다.

Markdown 파일은 보통 확장자가

```text
.md
```

이다.

예:

```text
study.md
python-study.md
README.md
database-study.md
```

확장자가 없거나 일반 텍스트 파일로 만들어지면 GitHub가 Markdown 문서로 렌더링하지 않을 수 있다.

따라서 GitHub에서 공부 내용을 기록할 때는

```text
2026-08-31-study.md
```

처럼 `.md` 확장자를 사용하는 것이 좋다.

# 12. AI 시대에도 데이터베이스를 공부해야 하는 이유

요즘은 ChatGPT나 Codex 같은 AI가 SQL도 작성해 준다.

예를 들어 AI에게 다음과 같은 작업을 요청할 수 있다.

```text
회원 테이블을 만들어줘.

두 테이블을 JOIN해줘.

ERD를 만들어줘.

인덱스를 추천해줘.
```

하지만 AI가 SQL을 작성했다고 해서 그 SQL이 반드시 올바른 것은 아니다.

SQL은 **오류 없이 실행되면서 잘못된 결과를 반환할 수도 있다.**

따라서 중요한 것은 SQL을 단순히 작성하는 것이 아니라

```text
왜 이 SQL을 사용하는가?

결과가 맞는가?

중복 데이터가 발생하지 않는가?

테이블 관계가 올바른가?

업무 요구사항과 결과가 일치하는가?
```

를 사람이 판단하는 것이다.

# 18. SQL

SQL은 데이터베이스에 명령을 내리는 언어이다.

기본적으로 많이 사용하는 명령은 다음과 같다.

```sql
CREATE
INSERT
SELECT
UPDATE
DELETE
```

각각의 역할은 다음과 같다.

| SQL    | 역할     |
| ------ | ------ |
| CREATE | 테이블 생성 |
| INSERT | 데이터 추가 |
| SELECT | 데이터 조회 |
| UPDATE | 데이터 수정 |
| DELETE | 데이터 삭제 |

이 다섯 가지는 데이터베이스 기본 학습에서 매우 중요하다.

# 19. ERD

ERD는

```text
Entity Relationship Diagram
```

의 약자이다.

데이터베이스의 테이블과 관계를 그림으로 표현한 것이다.

예:

```text
학생
  │
  │ 수강
  ↓
수강신청
  ↑
  │
강의
```

ERD를 만들 때 생각해야 하는 것은 다음과 같다.

```text
어떤 테이블이 필요한가?

한 행은 무엇을 의미하는가?

기본키는 무엇인가?

외래키는 무엇인가?

테이블끼리 어떤 관계가 있는가?
```

# 20. 기본키 PRIMARY KEY

PRIMARY KEY는 테이블의 각 행을 구분하기 위한 값이다.

예:

| student_id | name |
| ---------- | ---- |
| 1          | 철수   |
| 2          | 영희   |

여기서

```text
student_id
```

가 학생을 구분할 수 있는 PRIMARY KEY가 될 수 있다.

PRIMARY KEY는 중복될 수 없다.

---

# 21. 외래키 FOREIGN KEY

FOREIGN KEY는 다른 테이블의 데이터를 연결하기 위한 키이다.

예:

### students

| id | name |
| -- | ---- |
| 1  | 철수   |

### scores

| id | student_id | score |
| -- | ---------- | ----- |
| 1  | 1          | 95    |

여기서

```text
scores.student_id
```

가

```text
students.id
```

를 참조한다.

---

# 22. 데이터 무결성

데이터 무결성이란 데이터가 **정확하고 일관된 상태를 유지하는 것**을 의미한다.

이를 위해 다양한 제약조건을 사용한다.

```sql
PRIMARY KEY
FOREIGN KEY
NOT NULL
UNIQUE
CHECK
```

---

# 23. NOT NULL

값이 반드시 존재해야 할 때 사용한다.

예를 들어 학생 이름이 반드시 필요하다면

```sql
name VARCHAR(100) NOT NULL
```

처럼 지정할 수 있다.

---

# 24. UNIQUE

같은 값이 중복되면 안 될 때 사용한다.

예:

```text

사용자 아이디
사원번호
```

등이다.

---

# 25. CHECK

입력할 수 있는 값의 범위를 제한한다.

예를 들어 점수가

```text
0 ~ 100
```

이어야 한다면 CHECK 조건을 사용할 수 있다.

---

# 26. 정규화

정규화는 중복 데이터를 줄이고 데이터를 더 안정적으로 관리하기 위해 테이블을 적절하게 나누는 과정이다.

한 테이블에 모든 데이터를 저장하면 같은 정보가 반복될 가능성이 높아진다.

따라서

```text
학생
강의
수강신청
```

처럼 역할에 맞게 테이블을 분리한다.

---

# 27. JOIN

JOIN은 여러 테이블의 데이터를 연결해 조회할 때 사용한다.

예를 들어

```text
학생 테이블

+

수강신청 테이블

+

강의 테이블
```

을 연결하면

```text
어떤 학생이 어떤 강의를 듣고 있는가?
```

라는 질문에 답할 수 있다.

하지만 JOIN 조건이 잘못되면 행이 중복되어 합계나 개수가 틀릴 수 있기 때문에 반드시 결과를 검증해야 한다.

# 28. 집계

대표적인 집계 함수:

```sql
COUNT()
SUM()
AVG()
MAX()
MIN()
```

예:

```text
학생 수

평균 점수

총 매출

최고 점수
```

등을 계산할 때 사용할 수 있다.

---

# 29. 트랜잭션

여러 SQL 명령이 하나의 작업으로 처리되어야 할 때 사용한다.

대표적인 명령:

```sql
BEGIN
COMMIT
ROLLBACK
SAVEPOINT
```

예를 들어 계좌이체는

```text
A 계좌에서 돈 차감
        ↓
B 계좌에 돈 추가
```

두 작업이 모두 성공해야 한다.

한쪽만 성공하면 데이터가 잘못된다.

따라서 두 작업을 하나의 트랜잭션으로 관리한다.

---

# 30. COMMIT

작업을 데이터베이스에 최종 반영한다.

```text
SQL 실행
↓
문제 없음
↓
COMMIT
```

---

# 31. ROLLBACK

작업 중 문제가 발생하면 이전 상태로 되돌린다.

```text
SQL 실행
↓
```
