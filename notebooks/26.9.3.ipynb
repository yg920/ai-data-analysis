## 1. 오늘의 학습 목표

오늘은 본격적인 데이터 분석에 들어가기 전에 필요한 **Python 데이터 분석 개발 환경**을 구성하는 방법을 학습했다.

데이터 분석에서는 단순히 Python 코드만 작성한다고 끝나는 것이 아니다.

다음 환경들이 서로 올바르게 연결되어 있어야 한다.

```text
GitHub 저장소
    ↓
VS Code
    ↓
Python
    ↓
.venv 가상환경
    ↓
requirements.txt
    ↓
Jupyter Notebook
    ↓
샘플 데이터
    ↓
실행 결과 검증
```

환경이 제대로 연결되어 있지 않으면 같은 코드라도 여러 오류가 발생할 수 있다.

대표적으로 다음과 같은 문제가 있다.

- `ModuleNotFoundError`
- `FileNotFoundError`
- Notebook 커널 불일치
- 패키지를 설치했는데 `import` 실패
- 가상환경 활성화 오류
- 잘못된 Python 인터프리터 사용
- 데이터 파일 경로 오류

따라서 데이터 분석을 시작하기 전에 먼저 **내 코드가 어떤 Python에서 실행되는지 확인하는 습관**이 중요하다.

# 2. 환경 설정이 중요한 이유

데이터 분석 프로젝트에는 여러 종류의 파일과 환경이 함께 사용된다.

```text
Python 코드
CSV 데이터
Jupyter Notebook
Python 패키지
환경변수
분석 결과
Git 변경 이력
```

환경이 제대로 정리되어 있지 않으면 같은 프로젝트를 나중에 다시 실행하거나 다른 컴퓨터에서 실행하기 어려울 수 있다.

따라서 데이터 분석 프로젝트는 다음과 같은 형태로 구성하는 것이 좋다.

```text
프로젝트 폴더
+
프로젝트 전용 Python 환경
+
필요한 패키지
+
올바른 Notebook 커널
+
정확한 데이터 경로
```

즉 환경 설정은 단순히 프로그램을 설치하는 것이 아니라 **재현 가능한 작업 공간을 만드는 과정**이다.

# 8. 가상환경이란?

가상환경은 프로젝트마다 Python과 Python 패키지를 독립적으로 관리하기 위한 공간이다.

가상환경 이름으로 보통 다음을 사용한다.

```text
.venv
```

예를 들어 두 프로젝트가 있다고 생각해보자.

```text
프로젝트 A
└─ .venv
   ├─ pandas
   └─ numpy

프로젝트 B
└─ .venv
   ├─ Django
   └─ psycopg
```

각 프로젝트가 자신의 가상환경을 사용하면 필요한 패키지를 독립적으로 관리할 수 있다.

즉 다른 프로젝트의 패키지와 충돌할 가능성을 줄일 수 있다.

---

# 9. 프로젝트 전용 가상환경 생성

Windows에서 다음 명령을 실행한다.

```powershell
python -m venv .venv
```

환경에 따라 다음 명령을 사용할 수도 있다.

```powershell
py -m venv .venv
```

명령을 나누어서 보면 다음과 같다.

```text
python
→ Python 실행

-m venv
→ Python의 venv 모듈 실행

.venv
→ .venv라는 이름의 가상환경 생성
```

정상적으로 생성되면 프로젝트 안에 다음 폴더가 생긴다.

```text
.venv/
```

---

# 10. 가상환경 활성화

Windows PowerShell에서는 다음 명령을 사용한다.

```powershell
.\.venv\Scripts\Activate.ps1
```

정상적으로 활성화되면 터미널 앞부분에 다음과 같이 표시된다.

```text
(.venv) PS C:\dev\llm-data-analysis-course>
```

앞에 표시되는

```text
(.venv)
```

는 현재 터미널이 프로젝트 가상환경을 사용하고 있다는 뜻이다.

---

# 11. PowerShell 실행 정책 오류

PowerShell에서 가상환경을 활성화하려고 할 때 다음과 같은 오류가 발생할 수 있다.

```text
running scripts is disabled on this system
```

개인 학습 PC이고 조직 보안 정책에 문제가 없는 경우 현재 PowerShell 창에 대해서만 임시로 실행을 허용할 수 있다.

```powershell
Set-ExecutionPolicy -Scope Process -ExecutionPolicy Bypass
```

그다음 다시 실행한다.

```powershell
.\.venv\Scripts\Activate.ps1
```

`Process` 범위는 현재 PowerShell 창에서만 적용된다.

PowerShell 창을 닫으면 설정도 끝난다.

# 13. 오늘 직접 경험한 .venv 생성 오류

오늘 다음 명령으로 가상환경을 생성했다.

```powershell
python -m venv .venv
```

그런데 다음 오류가 발생했다.

```text
KeyboardInterrupt
```

Traceback을 확인했을 때 다음 과정에서 작업이 중단되었다.

```text
_setup_pip
ensurepip
```

즉 Python이 `.venv`를 만드는 과정에서 **pip를 준비하는 단계에서 작업이 중단된 것**을 확인했다.

---

# 14. 문제가 생긴 가상환경 삭제

가상환경 생성이 중간에 실패하면 불완전한 `.venv`가 남아 있을 수 있다.

PowerShell에서는 다음 명령으로 삭제할 수 있다.

```powershell
Remove-Item -Recurse -Force .venv
```

그다음 다시 가상환경 생성을 시도할 수 있다.

```powershell
python -m venv .venv
```

# 15. ensurepip 확인

```powershell
python -m ensurepip --upgrade
```

`ensurepip`는 Python에서 pip를 준비할 때 사용되는 기능이다.

# 30. ModuleNotFoundError

# 32. .env란?

`.env` 파일은 API Key나 비밀번호처럼 외부에 공개하면 안 되는 값을 저장할 때 사용한다.

예:

```env
OPENAI_API_KEY=실제_API_KEY
```

실제 API Key를 Python 코드나 Notebook에 직접 작성하지 않는 것이 좋다.

잘못된 예:

```python
OPENAI_API_KEY = "실제_API_KEY"
```

이 코드가 GitHub에 올라가면 Key가 공개될 수 있다.

---

# 33. .env.example

`.env.example`에는 실제 비밀 값을 저장하지 않는다.

어떤 환경변수가 필요한지만 알려준다.

예:

```env
OPENAI_API_KEY=
DATABASE_URL=
```

Windows에서는 다음과 같이 `.env.example`을 복사해 `.env`를 만들 수 있다.

```powershell
Copy-Item .env.example .env
```

그리고 자신의 실제 값을 `.env`에 작성한다.

# 34. .gitignore

`.gitignore`는 Git이 추적하지 않을 파일이나 폴더를 지정한다.

예:

```gitignore
.venv/
.env
__pycache__/
```

특히 `.env`에는 중요한 Key가 들어갈 수 있기 때문에 GitHub에 올라가지 않도록 해야 한다.

해당 Key를 **폐기하거나 재발급**해야 한다.
