# 1. AI 시대에 데이터베이스를 왜 배워야 할까?

AI에게 요청하면 SQL이나 테이블 구조를 매우 빠르게 만들어낼 수 있다.

예를 들어 다음과 같은 요청이 가능하다.

```text
학생 정보를 저장할 PostgreSQL 테이블을 만들어 주세요.

학생별 질문 수를 계산하는 SQL도 작성해 주세요.
```

AI는 SQL 초안을 빠르게 작성할 수 있지만 실제 업무 규칙까지 자동으로 정확하게 이해하는 것은 아니다.

예를 들어 다음과 같은 문제는 사람이 판단해야 한다.

* 학생 이메일은 중복될 수 있는가?
* 탈퇴 학생의 데이터는 삭제해야 하는가?
* 질문을 작성하지 않은 학생도 통계에 포함해야 하는가?
* 어떤 데이터가 개인정보인가?
* 어떤 사람이 어떤 데이터를 조회할 수 있는가?

즉,

> AI는 SQL을 만들어 줄 수 있지만
> 데이터의 의미와 업무 규칙을 결정하는 것은 사람이다.

# 2. SQL이 실행된다고 해서 정답은 아니다

오늘 배운 내용 중 매우 중요한 개념이다.

```text
실행되는 SQL ≠ 올바른 SQL
```

SQL 문법에 오류가 없어서 실행된다고 하더라도 실제 요구사항과 다른 결과가 나올 수 있다.

예를 들어 전체 학생 수를 알고 싶다고 하자.

```sql
SELECT COUNT(*) AS student_count
FROM students AS s
INNER JOIN questions AS q
    ON q.student_id = s.id;
```

문법적으로는 정상적인 SQL이다.

하지만 `INNER JOIN`을 사용하면 질문을 작성하지 않은 학생은 결과에서 사라질 수 있다.

또한 한 학생이 질문을 여러 개 작성했다면 그 학생이 여러 행으로 나타날 수도 있다.

전체 학생 수만 필요하다면 오히려 다음 SQL이 더 적절할 수 있다.

```sql
SELECT COUNT(*) AS student_count
FROM students;
```

따라서 SQL을 확인할 때 다음 질문을 해야 한다.

1. 무엇을 조회하려는가?
2. 결과의 한 행은 무엇을 의미하는가?
3. 누락되는 데이터는 없는가?
4. 중복되는 데이터는 없는가?
5. 예상 결과와 실제 결과가 같은가?

# 3. 데이터베이스 기본 용어

오늘 여러 데이터베이스 용어도 확인했다.

## JOIN

서로 관련된 여러 테이블의 데이터를 연결해서 조회하는 방법이다.

예를 들어

```text
학생
↓
수강신청
↓
강의
```

처럼 서로 다른 테이블의 데이터를 연결할 때 사용할 수 있다.

---

## NULL

값이 없거나 아직 입력되지 않은 상태를 의미한다.

중요한 점은

```text
NULL ≠ 0
```

이라는 것이다.

예를 들어 시험 점수에서

```text
0점
```

은 시험을 봤지만 0점을 받은 것이다.

반면

```text
NULL
```

은 아직 점수가 입력되지 않았을 수도 있다.

따라서 무조건 NULL을 0으로 변환해서는 안 된다.

---

## 기본키 Primary Key

테이블의 각 행을 구분하기 위한 값이다.

예:

```text
student_id
order_id
product_id
```

---

## 외래키 Foreign Key

다른 테이블의 데이터를 참조하기 위한 값이다.

예:

```text
orders.customer_id
```

가

```text
customers.customer_id
```

를 참조할 수 있다.

---

## 제약조건 Constraint

잘못된 데이터가 저장되는 것을 방지하는 규칙이다.

예를 들어

```text
이메일 중복 금지
점수는 0~100점
필수 데이터는 NULL 금지
```

같은 규칙을 데이터베이스에 적용할 수 있다.

---

## 트랜잭션 Transaction

여러 데이터 변경 작업을 하나의 작업 단위로 처리하는 방법이다.

예를 들어 계좌 이체라면

```text
A 계좌에서 10,000원 차감
+
B 계좌에 10,000원 추가
```

두 작업이 모두 성공하거나 모두 실패해야 한다.

---

## 인덱스 Index

데이터를 더 빠르게 찾기 위해 사용하는 구조이다.

책의 목차나 색인과 비슷하다고 생각하면 된다.

---

# 4. 잘못된 데이터는 잘못된 분석을 만든다

데이터 분석에서는 입력 데이터가 매우 중요하다.

전체 과정은 다음처럼 연결될 수 있다.

```text
잘못된 데이터
↓
잘못된 SQL 결과
↓
잘못된 AI 분석
↓
잘못된 의사결정
```

예를 들어

### 중복 주문

```text
실제 주문 금액보다 크게 계산
↓
매출을 과대평가
```

### 잘못된 JOIN

```text
일부 고객 누락
↓
전체 고객 현황을 잘못 판단
```

### 잘못된 필터

```text
취소 주문까지 매출에 포함
↓
매출 과대 계산
```

### 오래된 상태값

```text
과거 상태
↓
현재 상태처럼 분석
```

따라서 데이터 분석을 할 때는 데이터 품질과 계산 기준을 반드시 확인해야 한다.

---

# 5. 데이터는 서로 관계를 가진다

현실의 서비스 데이터는 하나의 테이블만으로 구성되는 경우가 거의 없다.

예를 들어 쇼핑몰은

```text
회원
↓
주문
↓
주문 상세
↓
상품
```

처럼 여러 데이터가 연결된다.

온라인 교육 서비스라면

```text
학생
↓
수강신청
↓
강의
```

처럼 표현할 수 있다.

데이터베이스 설계는 결국

```text
현실의 업무
↓
데이터
↓
데이터 관계
↓
테이블
↓
제약조건
```

으로 바꾸는 과정이라고 볼 수 있다.

# 6. 파일 / 스프레드시트 / DBMS 선택

모든 데이터를 데이터베이스에 저장해야 하는 것은 아니다.

상황에 따라 파일이나 Excel 같은 스프레드시트가 더 편할 수도 있다.

## 파일이 적합한 경우

```text
개인 메모
설정 파일
간단한 데이터
```

## 스프레드시트가 적합한 경우

```text
작은 명단
간단한 일정
소규모 통계
```

## DBMS가 적합한 경우

다음 조건이 많을수록 데이터베이스가 필요할 가능성이 높다.

* 데이터가 계속 증가한다.
* 데이터가 자주 변경된다.
* 여러 사용자가 동시에 사용한다.
* 데이터 사이의 관계가 중요하다.
* 검색과 집계를 반복한다.
* 권한 관리가 필요하다.
* 변경 이력이 중요하다.
* 백업과 복구가 중요하다.
* 웹이나 앱에서 지속적으로 사용한다.

# 8. 데이터 분석은 코드가 아니라 질문에서 시작한다

오늘 AI 데이터 분석 수업에서 가장 중요했던 내용 중 하나이다.

데이터 분석은 바로 Python 코드를 작성하면서 시작하는 것이 아니다.

먼저

```text
무엇이 궁금한가?
```

를 결정해야 한다.

예를 들어

```text
요즘 매출이 줄어든 것 같은데 왜 그런가요?
```

라는 질문은 사람에게는 이해되지만 분석하기에는 너무 모호하다.

다음 내용을 먼저 정의해야 한다.

```text
요즘 = 언제?

매출 = 어떤 금액?

완료 주문만 볼 것인가?

취소 주문은 제외할 것인가?

전월과 비교할 것인가?

전년과 비교할 것인가?
```

# 9. 분석 가능한 질문으로 바꾸기

예를 들어

```text
어떤 상품이 잘 팔리나요?
```

라는 질문이 있다고 하자.

여기서

```text
잘 팔린다
```

는 여러 의미가 될 수 있다.

* 판매 수량이 많다.
* 판매 금액이 크다.
* 주문 건수가 많다.
* 반복 구매가 많다.

따라서 질문을 구체적으로 바꿀 수 있다.

```text
상품별 총 판매 수량과
완료 주문 기준 판매 금액을 계산했을 때
상위 상품은 어떻게 다른가?
```

이렇게 질문을 구체화하면 어떤 데이터가 필요한지 결정할 수 있다.

예:

```text
quantity
unit_price
product_id
product_name
order_status
```

# 11. 데이터 분석 전체 흐름

일반적인 데이터 분석 과정은 다음과 같다.

```text
문제 정의
↓
데이터 수집
↓
데이터 구조 이해
↓
전처리
↓
EDA
↓
시각화
↓
해석
↓
보고서
↓
의사결정
```

AI를 사용하더라도 이 과정 자체가 없어지는 것은 아니다.

오히려 AI가 빠르게 많은 결과를 만들기 때문에 검증 과정이 더욱 중요하다.

# 12. LLM이 잘 도와주는 일

LLM은 데이터 분석에서 좋은 보조 도구가 될 수 있다.

예를 들어

* 분석 질문 후보 만들기
* 필요한 컬럼 제안
* 데이터 품질 확인 항목 제안
* pandas 코드 초안 작성
* 오류 메시지 설명
* 그래프 해석 초안 작성
* 보고서 구조 작성
* 검증 체크리스트 작성

등을 도와줄 수 있다.

하지만

```text
자연스러운 AI 답변
≠
검증된 분석
```

이다.

# 13. 사람과 AI의 역할

오늘 배운 데이터 분석의 중요한 흐름이다.

```text
사람이 질문 정의
↓
LLM이 초안 작성
↓
실제 데이터로 검증
↓
사람이 수정
↓
사람이 최종 판단
```

즉 AI에게 모든 분석을 맡기는 것이 아니라

```text
검토할 초안을 빠르게 만드는 도구
```

로 사용하는 것이 좋은 방법이다.

# 14. LLM이 대신할 수 없는 것

LLM은 다음 내용을 자동으로 보장하지 않는다.

* 데이터가 정확한가?
* 실제 컬럼명이 존재하는가?
* 데이터 타입이 맞는가?
* 결측치 처리가 적절한가?
* 중복 데이터가 있는가?
* 이상값 처리가 맞는가?
* 업무 기준을 정확하게 이해했는가?
* 데이터 누수가 없는가?
* 분석 결과가 충분히 검증됐는가?
* 개인정보가 안전하게 관리되고 있는가?

예를 들어 AI가

```text
고객 연령대별 구매 금액을 분석하세요.
```

라고 추천했다고 하자.

그런데 실제 데이터에

```text
age
```

컬럼이 없다면 분석할 수 없다.

따라서 항상

```text
AI 제안
↓
실제 컬럼 확인
↓
데이터 확인
↓
코드 작성
↓
실행
↓
결과 검증
```

과정이 필요하다.

# 16. 쇼핑몰 데이터 관계

고객과 주문의 관계

```text
customers.customer_id
        ↓
orders.customer_id
```

주문과 주문 상세의 관계

```text
orders.order_id
        ↓
order_items.order_id
```

```text
products.product_id
        ↓
order_items.product_id
```

```text
고객별 구매 금액
상품별 판매 수량
카테고리별 주문 금액
월별 주문 추이
```

# 17. EDA와 머신러닝

## EDA

EDA는 이미 존재하는 데이터에서 패턴을 찾는 과정이다.

예:

# 18. 데이터 누수 Data Leakage

머신러닝에서는 매우 중요한 개념이다.

예를 들어 주문 취소 여부를 예측한다고 하자.

그런데 입력 데이터에

```text
order_status
```

를 넣었다.

`order_status`에 이미 취소 여부가 들어 있다면 정답을 미리 모델에게 알려준 것이다.

이를

```text
데이터 누수
Data Leakage
```

라고 한다.

따라서

```text
높은 모델 정확도
≠
좋은 모델
```

이다.

# 24. Django Static 파일

Django에서 이미지나 CSS 같은 정적 파일을 관리할 때 `static` 폴더를 사용한다.

오늘 사용한 구조는 다음과 비슷하다.

```text
expense/
│
├─ static/
│  └─ expense/
│     └─ background.png
│
├─ templates/
│
├─ models.py
├─ views.py
└─ urls.py
```

Static 파일 위치를 확인하기 위해 다음 명령어도 사용했다.

```bash
python manage.py findstatic expense/background.png --verbosity 2
```

이 명령을 통해 Django가 실제로 해당 이미지 파일을 찾을 수 있는지 확인할 수 있었다.

# 25. Django 배경 이미지 적용

HTML에서 Django Static 파일을 사용하려면 먼저

```django
{% load static %}
```

을 작성한다.

그리고 이미지 경로를 다음처럼 사용할 수 있다.

```django
{% static 'expense/background.png' %}
```

이를 CSS 배경 이미지와 함께 사용할 수 있다.

예:

```css
body {
    background-image: url("{% static 'expense/background.png' %}");
}
```

이 과정에서

```text
이미지가 안 나옴
↓
파일 위치 확인
↓
findstatic 확인
↓
Static 경로 수정
↓
이미지 표시 성공
```

과정을 경험했다.

# 29. Git 기본 명령어

# 30. GitHub Repository 연결

GitHub Repository를 연결할 때 다음 명령어를 사용할 수 있다.

```bash
git remote add origin https://github.com/사용자명/저장소명.git
```

현재 Remote 확인

```bash
git remote -v
```

Remote 주소를 잘못 등록했다면 다음처럼 수정할 수 있다.

```bash
git remote set-url origin https://github.com/사용자명/저장소명.git
```

---

# 31. GitHub Push

GitHub로 프로젝트를 올릴 때

```bash
git push -u origin main
```

을 사용할 수 있다.

오늘 다음 메시지도 확인했다.

```text
branch 'main' set up to track 'origin/main'.
Everything up-to-date
```

이 메시지는 로컬 `main` Branch와 GitHub의 `main` Branch가 연결되어 있으며 현재 추가로 Push할 변경 사항이 없다는 의미이다.

# 32. Repository not found 오류

GitHub 업로드 과정에서 다음 오류를 경험했다.

```text
fatal: repository '...' not found
```

이 오류가 발생할 때 확인할 항목은

```text
Repository 주소가 정확한가?
Repository가 실제로 존재하는가?
GitHub 로그인 계정이 맞는가?
Repository 접근 권한이 있는가?
Remote 주소가 잘못 등록되지 않았는가?
```

등이다.

특히 GitHub URL을 잘못 작성하면 Repository를 찾을 수 없다는 오류가 발생할 수 있다.

와 같은 데이터 분석을 할 수 있다.

그리고 개발한 코드는

```text
Git
↓
GitHub
```

로 관리한다.

# 34. 오늘 가장 중요한 개념

오늘 학습 내용을 한 문장으로 정리하면 다음과 같다.

> AI가 코드를 빠르게 만들어 줄 수는 있지만
> 문제를 정의하고 데이터의 의미를 이해하며
> 결과가 올바른지 검증하는 역할은 사람이 해야 한다.

데이터베이스에서도

```text
SQL 실행 성공
≠
올바른 결과
```

이고,

데이터 분석에서도

```text
AI가 자연스럽게 설명함
≠
정확한 분석
```

이며,

머신러닝에서도

```text
높은 정확도
≠
좋은 모델
```

이다.

따라서 앞으로 가장 중요하게 가져가야 할 습관은

```text
질문
↓
예상
↓
실행
↓
결과 확인
↓
검증
↓
수정
```

이다.

# ✅ 오늘의 최종 복습

오늘 배운 내용을 간단히 다시 정리하면 다음과 같다.

* AI는 SQL과 코드의 초안을 빠르게 만들 수 있다.
* AI가 만든 결과는 반드시 사람이 검증해야 한다.
* SQL이 실행된다고 결과가 정확한 것은 아니다.
* JOIN은 테이블을 연결하지만 데이터 누락이나 중복이 발생할 수 있다.
* NULL과 0은 서로 다른 의미이다.
* 기본키와 외래키를 통해 데이터 관계를 표현한다.
* 데이터베이스는 현실의 업무 구조를 데이터와 관계로 표현한다.
* 데이터 분석은 코드보다 질문 정의가 먼저다.
* 분석 질문은 대상·기준·비교·목적을 명확히 해야 한다.
* 데이터 분석은 문제 정의 → 데이터 → 전처리 → EDA → 시각화 → 해석 순서로 진행된다.
* LLM은 분석을 도와주는 도구이지 최종 판단자가 아니다.
* EDA는 기존 데이터의 패턴을 탐색한다.
* 머신러닝은 새로운 값을 예측한다.
* 데이터 누수를 조심해야 한다.
* Prompt와 AI 활용 과정을 기록하면 분석을 검증하기 쉽다.
* 개인정보와 API Key는 외부에 노출하면 안 된다.
* `.env`와 `.gitignore`를 이용해 Secret을 보호할 수 있다.
* Django에서는 `static` 폴더를 이용해 이미지와 정적 파일을 관리한다.
* `findstatic` 명령으로 Static 파일 위치를 확인할 수 있다.
* Django로 Python 프로그램을 웹 서비스 형태로 확장할 수 있다.
* Git은 코드의 변경 이력을 관리한다.
* GitHub는 Git Repository를 원격으로 저장하고 공유할 수 있다.
* `git add → commit → push`가 기본적인 Git 작업 흐름이다.
* GitHub Remote 주소가 틀리면 `repository not found` 오류가 발생할 수 있다.
* 개발, 데이터베이스, 데이터 분석, AI는 결국 서로 연결되는 기술이다.
