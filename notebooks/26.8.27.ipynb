## 1. 함수란?

함수는 **관련된 코드를 하나의 이름으로 묶어서 필요할 때 다시 사용하는 방법**이다.

함수의 기본적인 형태는 다음과 같다.

```python
def 함수이름():
    실행할 코드
```

예제:

```python
def say_hello():
    print("Hello, Python!")
```

여기까지만 작성하면 함수를 **정의**한 것이지 실행한 것은 아니다.

실제로 실행하려면 함수를 호출해야 한다.

```python
say_hello()
```

### 핵심

```text
def say_hello():
    → 함수 정의

say_hello()
    → 함수 호출
```

즉,

> 함수를 만든 것과 함수를 실행한 것은 다르다.

# 5. return

함수는 값을 받아 처리한 다음 **결과를 함수 밖으로 돌려줄 수 있다.**

이때 사용하는 것이 `return`이다.

```python
def add(a, b):
    return a + b
```

호출:

```python
total = add(10, 20)

print(total)
```

실행 과정은 다음과 같다.

```text
add(10, 20)

↓
a = 10
b = 20

↓
a + b

↓
30

↓
return 30

↓
total = 30
```

### 핵심

> `return`은 함수가 처리한 결과를 함수 밖으로 돌려준다.

반환된 값은 변수에 저장해서 다시 사용할 수도 있다.

# 6. return에서 `=`을 사용하지 않는다

학습하면서 다음과 같은 실수를 했다.

```python
return = score1 + score2
```

이것은 잘못된 문법이다.

`return`은 변수에 값을 저장하는 것이 아니기 때문에 `=`을 사용하지 않는다.

올바른 형태:

```python
return score1 + score2
```

# 7. print()와 return의 차이

둘은 비슷해 보이지만 역할이 다르다.

### print()

```python
def add_print(a, b):
    print(a + b)
```

`print()`는 결과를 **화면에 보여준다.**

### return

```python
def add_return(a, b):
    return a + b
```

`return`은 결과를 **함수 밖으로 전달한다.**

따라서:

```python
result = add_return(10, 20)
```

처럼 반환된 결과를 저장할 수 있다.

### 선택 기준

```text
사용자에게 보여주기
        ↓
      print()

다른 코드에서 결과 사용하기
        ↓
      return
```

```python
return score1 + score2
```

라고 작성해야 한다.

# 11. 사칙연산 함수 연습

사칙연산은 다음 네 가지 기본 연산이다.

```text
+  덧셈
-  뺄셈
*  곱셈
/  나눗셈
```

예:

```text
20 + 10 = 30
20 - 10 = 10
20 * 10 = 200
20 / 10 = 2.0
```

여러 값을 반환하는 기능을 이용하면 사칙연산 결과를 한 번에 반환하는 함수도 만들 수 있다.

```python
def calc(score1, score2):
    return (
        score1 + score2,
        score1 - score2,
        score1 * score2,
        score1 / score2
    )
```

# 16. 지역 변수

함수 안에서 만들어진 변수는 함수 안에서 사용하는 **지역 변수**가 된다.

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total
```

여기서:

```text
price, quantity
→ 함수가 받은 값

total
→ 함수 안의 지역 변수
```

함수 밖에서는:

```python
result = calculate_total(12000, 3)
```

처럼 반환값을 다른 변수에 저장할 수 있다.

```text
total
→ 함수 내부

result
→ 함수 외부
```

둘은 서로 다른 변수다.

# 17. 함수 안에서 조건문 사용

함수 안에서도 `if`, `elif`, `else`를 사용할 수 있다.

```python
def get_grade(score):

    if score >= 90:
        return "A"

    elif score >= 80:
        return "B"

    else:
        return "C"
```

즉 기존에 배운 조건문을 **함수라는 하나의 역할로 묶을 수 있다.**

# 18. 함수 안에서 반복문 사용

반복문도 함수 안에서 사용할 수 있다.

```python
def calculate_total(scores):

    total = 0

    for score in scores:
        total += score

    return total
```

호출:

```python
scores = [80, 90, 100]

result = calculate_total(scores)

print(result)
```

함수 안에 반복문이 들어갔다고 새로운 반복 규칙이 생기는 것은 아니다.

> 지금까지 배운 문법들을 함수라는 하나의 역할 안에 묶는 것이다.

```text
import math
→ math 모듈 전체 사용

from math import sqrt
→ math에서 sqrt만 가져옴

import statistics as stats
→ statistics를 stats라는 별칭으로 사용
```

활성화되면 터미널 앞에 보통:

```text
(.venv)
```

가 나타난다.

예:

```text
(.venv) PS C:\project>
```

가상환경에서 나오려면:

```bash
deactivate
```

를 사용한다.

---

# 33. (.venv)만 믿으면 안 된다

터미널에 `(.venv)`가 보인다고 무조건 올바른 환경이라고 생각하지 않는다.

현재 Python 위치를 확인할 수 있다.

```bash
python -c "import sys; print(sys.executable)"
```

pip 위치도 확인한다.

```bash
python -m pip --version
```

### 가장 중요한 질문

> 패키지를 설치한 Python과 코드를 실행하는 Python이 같은 환경인가?

---

# 34. VS Code Interpreter

가상환경을 만들었다고 VS Code가 무조건 해당 Python을 사용하는 것은 아니다.

VS Code에서:

```text
Ctrl + Shift + P
```

를 누른 후:

```text
Python: Select Interpreter
```

를 선택한다.

그리고 프로젝트의 `.venv`에 해당하는 Python을 선택한다.

선택 후 새 터미널을 열고 Python 경로를 다시 확인하는 것이 좋다.

---

# 35. ipykernel 설치 메시지

VS Code에서 Jupyter 셀을 실행할 때 다음과 비슷한 메시지를 볼 수 있다.

```text
Running cells with '.venv ...'
requires the ipykernel package.
```

이것은 Python을 다시 설치하라는 뜻이 아니다.

현재 `.venv` 환경에서 Jupyter 셀을 실행하는 데 필요한 **ipykernel 패키지가 없다는 의미**다.

즉 가상환경 안에 필요한 패키지를 추가하는 과정이다.

---

# 36. .venv는 Git에 올리지 않는다

`.venv` 안에는 설치된 패키지와 실행환경 파일이 많이 들어 있다.

그리고 다시 만들 수 있기 때문에 일반적으로 Git에 올리지 않는다.

`.gitignore`에:

```text
.venv/
```

를 추가한다.

대신 필요한 패키지 정보를 `requirements.txt`에 기록한다.

---

# 37. requirements.txt

현재 환경에 설치된 패키지를 확인한다.

```bash
python -m pip freeze
```

파일로 저장:

```bash
python -m pip freeze > requirements.txt
```

그러면 현재 프로젝트가 어떤 패키지를 사용하는지 기록할 수 있다.

다른 환경에서는:

```bash
python -m pip install -r requirements.txt
```

로 다시 설치할 수 있다.

### 핵심

```text
.venv
→ 실제 설치 환경

requirements.txt
→ 환경을 다시 만들기 위한 패키지 목록
```

---

# 38. 프로젝트 환경을 다시 만드는 흐름

```text
프로젝트 코드
+
requirements.txt

        ↓

새로운 .venv 생성

        ↓

requirements.txt의 패키지 설치

        ↓

프로젝트 실행 환경 재구성
```

즉 `.venv` 자체를 다른 사람에게 보내기보다 **필요한 패키지 목록을 전달해서 환경을 다시 만드는 방식**을 사용할 수 있다.

---

# 39. ModuleNotFoundError

다음과 같은 오류가 발생할 수 있다.

```text
ModuleNotFoundError
```

이때 무조건 패키지를 다시 설치하는 것이 아니라 순서대로 확인한다.

```text
1. import 이름이 맞는가?

        ↓

2. 직접 만든 모듈이라면
   파일 위치가 맞는가?

        ↓

3. 외부 패키지가
   현재 환경에 설치되어 있는가?

        ↓

4. 현재 Python은
   어느 경로인가?

        ↓

5. VS Code도 같은
   Interpreter를 사용하는가?
```

---

# 40. 설치했는데 import가 안 되는 이유

예를 들어:

```text
터미널 A의 Python
        ↓
requests 설치

VS Code
        ↓
다른 Python으로 실행
```

이라면 패키지를 분명 설치했어도 VS Code에서는 찾지 못할 수 있다.

그래서 다음을 확인한다.

```bash
python -m pip show requests
```

그리고:

```bash
python -c "import sys; print(sys.executable)"
```

마지막으로 VS Code의:

```text
Python: Select Interpreter
```

도 확인한다.

### 중요한 개념

> "패키지가 설치되어 있는가?"와
> "현재 실행 중인 Python에 패키지가 설치되어 있는가?"는 서로 다른 질문이다.

---

# 41. 함수에서 모듈로 확장하기

이번 학습에서 Chapter 14와 Chapter 15는 서로 연결된다.

처음에는:

```python
def calculate_total(price, quantity):
    return price * quantity
```

처럼 함수를 만들었다.

함수가 많아지면 관련 함수들을:

```text
price.py
```

같은 파일로 옮길 수 있다.

더 커지면:

```text
shop/
├── __init__.py
├── price.py
└── delivery.py
```

처럼 패키지로 구성할 수 있다.

그리고 프로젝트별 실행 환경은:

```text
.venv/
```

로 분리한다.

즉 전체적인 발전 과정은:

```text
코드
 ↓
함수
 ↓
모듈
 ↓
패키지
 ↓
프로젝트 구조
 ↓
가상환경
```

으로 이해할 수 있다.

# 43. 어제 배운 내용 최종 정리

## 함수

```text
def
→ 함수 정의

함수이름()
→ 함수 호출

매개변수
→ 함수가 값을 받을 자리

인수
→ 실제 전달하는 값

return
→ 처리 결과를 함수 밖으로 전달

print()
→ 화면에 결과 출력
```

여러 값을 반환하면 튜플 형태로 받을 수 있고 언패킹할 수도 있다.

## pip

```text
외부 패키지 설치
```

# ⭐ 가장 중요하게 기억할 흐름

```text
함수
│
│ 하나의 역할을 묶는다
↓
모듈
│
│ 관련 함수들을 .py 파일로 묶는다
↓
패키지
│
│ 관련 모듈들을 폴더로 묶는다
↓
프로젝트
│
│
├── Python 코드
├── 패키지
├── requirements.txt
│
└── .venv
     ↓
프로젝트 전용 실행 환경
```

그리고 환경 오류가 발생했을 때는:

```text
이름 확인
   ↓
파일 위치 확인
   ↓
패키지 설치 확인
   ↓
Python 경로 확인
   ↓
VS Code Interpreter 확인
```

순서로 접근한다.
