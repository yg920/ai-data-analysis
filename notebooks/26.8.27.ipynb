# 1. 파일 읽기

Python에서는 `open()`을 사용해 파일을 열 수 있다.

```python
with open("memo.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line)
```

`r`은 **read(읽기)**를 의미한다.

```python
open("memo.txt", "r")
```

즉,

> memo.txt 파일을 읽기 모드로 연다.

### `with`를 사용하는 이유

```python
with open(...) as file:
```

파일을 사용한 뒤 자동으로 닫아주기 때문에 편리하고 안전하다.

# 3. 직접 줄 번호 붙이기

오늘 처음 작성했던 코드:

```python
with open("memo.txt", "r", encoding="utf-8") as file:

    leng = 1

    for line in file:
        print(f"{leng}번째 라인 {line.split()}")
        leng += 1
```

### 동작 순서

```text
leng = 1

첫 번째 줄 읽기
→ 1번째 라인 출력
→ leng += 1

두 번째 줄 읽기
→ 2번째 라인 출력
→ leng += 1

...
```

### `leng += 1`

다음과 같은 의미다.

```python
leng = leng + 1
```

즉, 기존 숫자에 1을 더한다.

파일에 새로운 내용을 저장할 때 `"w"` 모드를 사용할 수 있다.

# 9. 마지막 줄에 `\n` 넣지 않기

과제 조건:

```text
새로운 내용을 추가합니다. 1번째
새로운 내용을 추가합니다. 2번째
...
새로운 내용을 추가합니다. 10번째
```

단,

> 마지막 10번째 줄에는 `\n`을 추가하지 않는다.

작성한 코드:

```python
with open("result.txt", "a", encoding="utf-8") as file:

    for num in range(1, 11):

        if num < 10:
            file.write(f"새로운 내용을 추가합니다. {num}번째\n")

        else:
            file.write(f"새로운 내용을 추가합니다. {num}번째")
```

### 조건을 한국어로 읽기

```python
if num < 10:
```

> num이 10보다 작다면 줄바꿈을 넣는다.

```python
else:
```

> 그렇지 않다면, 즉 10이라면 줄바꿈을 넣지 않는다.

# 20. `try-except`

예상 가능한 실행 중 예외를 처리할 때 사용할 수 있다.

기본 구조:

```python
try:
    문제가 발생할 수 있는 코드

except 예외이름:
    문제가 발생했을 때 실행할 코드
```

예:

```python
try:
    age = int(input("나이를 입력하세요: "))
    print(age)

except ValueError:
    print("숫자로 입력해 주세요.")
```

### 역할

`try`

> 일단 이 코드를 실행해 본다.

`except`

> 지정한 문제가 발생하면 프로그램을 바로 끝내는 대신 대응한다.

# 22. 여러 개의 `except`

하나의 코드에서도 여러 종류의 문제가 발생할 수 있다.

```python
try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except ValueError:
    print("정수를 입력해 주세요.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
```

입력에 따라:

```text
abc → ValueError
0   → ZeroDivisionError
5   → 정상 실행
```

처럼 서로 다른 처리를 할 수 있다.

# 23. `else`와 `finally`

예외 처리에는 `else`, `finally`도 사용할 수 있다.

```python
try:
    number = int(input("정수: "))

except ValueError:
    print("정수가 아닙니다.")

else:
    print(f"입력 성공: {number}")

finally:
    print("입력 처리를 마쳤습니다.")
```

### 실행 조건

`try`
→ 문제가 발생할 수 있는 작업

`except`
→ 지정한 예외가 발생했을 때

`else`
→ `try`에서 예외가 발생하지 않았을 때

`finally`
→ 성공하거나 실패하거나 관계없이 마지막에 실행

초보 단계에서는 먼저 `try-except`부터 익히는 것이 중요하다.

## 들여쓰기

Python에서는 코드가 어느 블록에 속하는지 들여쓰기로 표현한다.

```python
with ...:
    for ...:
        if ...:
            file.write(...)
        else:
            file.write(...)
```

`with → for → if`처럼 안쪽으로 들어갈수록 들여쓰기 단계도 증가한다.

오늘 배운 내용을 전체 흐름으로 보면:

```text
파일 열기
   ↓
with open()
   ↓
r / w / a 모드 선택
   ↓
파일 한 줄씩 읽기
   ↓
for line in file
   ↓
strip()으로 정리
   ↓
split()으로 데이터 분리
   ↓
for + range()로 반복
   ↓
if / else로 조건 처리
   ↓
file.write()로 저장
   ↓
Path로 파일 위치 확인
   ↓
오류 발생
   ↓
트레이스백 확인
   ↓
예외 종류 파악
   ↓
필요한 경우 try-except로 대응
```
