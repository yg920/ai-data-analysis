# 5. PK와 FK를 실제로 관찰

## 5-1. 정상 데이터의 관계 읽기

다음 SQL 결과를 보고 작성합니다.

```sql
SELECT
    e.id AS enrollment_id,
    s.name AS student_name,
    c.title AS course_title,
    e.status
FROM ch02_enrollments AS e
JOIN ch02_students AS s
    ON s.id = e.student_id
JOIN ch02_courses AS c
    ON c.id = e.course_id
ORDER BY e.id;
```

```text
한 행이 의미하는 것: 한 학생의 수강하고 있는 한 건

같은 student_id가 여러 enrollment 행에서 반복될 수 있는 이유: 학생 한명은 여러 강의를 수강 가능

같은 course_id가 여러 enrollment 행에서 반복될 수 있는 이유: 한 강의를 여러 학생이 수강하기 때문
```

## 5-2. 기본키 중복 오류 관찰

중복 PK 입력을 시도한 결과: id = 1로 추가하려 햇더니 오류가 발생

```text
실행 성공 / 실패: 실패
오류 메시지에서 확인한 핵심 단어: 키가 이미 잇습니다.
왜 실패했다고 생각하는가: PK는 절대 중복될 수 없음
```

## 5-3. 존재하지 않는 학생을 참조하는 FK 오류 관찰

존재하지 않는 `student_id`를 사용한 수강신청 입력 결과:

```text
실행 성공 / 실패: 실패
오류 메시지에서 확인한 핵심 단어: 참조키(foreign key) 제약 조건을 위배했습니다
왜 실패했다고 생각하는가: student_id는 ch02_students의 id를 참조하는 FK인데,
존재하지 않는 학생 id를 입력했기 때문에 실패
```

## 5-4. PK와 FK의 차이 정리

```text
PK는 데이터베이스에서 고유갑승로 식별하기 위한 키이다.

FK는 다른 테이블에 관게를 연결 하기 위한 키이다.

FK 값이 여러 행에서 반복될 수 있는 이유는
하나의 부모 행을 여러 자식이 참조할 수 있기 때문이다.
```

```
개인 지출 관리 서비스의 데이터베이스 구조를 검토해 주세요.

현재 생각한 테이블은 users, expenses, categories입니다.

- users: 사용자 정보
- expenses: 지출 내역
- categories: 지출 카테고리

FK 후보는
expenses.user_id → users.user_id
expenses.category_id → categories.category_id 입니다.

이 구조에서 빠진 부분이나 개선할 점이 있는지 질문 중심으로 검토해 주세요.
아직 정하지 않은 정책은 임의로 확정하지 말아 주세요.
```

## 9-2. AI가 질문한 내용 중 유용했던 것

```text
1. 한 사용자가 여러 지출 내역을 가질 수 있는 구조인지 확인한 질문이 유용했다.
2. 지출 카테고리를 반드시 선택해야 하는지, 선택 사항으로 둘지 묻는 질문이 유용했다.
3. 카테고리를 사용자가 직접 추가할 수 있게 할지, 미리 정해진 값만 사용할지 묻는 질문이 유용했다.
```

## 9-3. AI가 너무 빨리 결정한 내용 또는 내가 보류한 내용

```text
1. 지출 내역 삭제 방식을 실제 삭제로 할지, 삭제 상태만 표시하는 방식으로 할지는 아직 보류했다.
2. 카테고리를 사용자가 직접 추가할 수 있게 할지, 미리 정해진 카테고리만 사용하게 할지는 아직 보류했다.
```

## 9-4. 검토 후 수정한 구조

| 수정 전                  | 수정 후                                              | 수정 이유                           |
| --------------------- | ------------------------------------------------- | ------------------------------- |
| 지출 내역에 사용자 정보까지 함께 저장 | users 테이블과 expenses 테이블을 분리하고 user_id로 연결         | 사용자 정보가 반복 저장되는 것을 줄이기 위해       |
| 지출 내역에 카테고리 이름을 직접 저장 | categories 테이블을 만들고 category_id로 연결               | 카테고리 값을 일관되게 관리하기 위해            |
| 테이블 간 관계가 명확하지 않음     | expenses.user_id와 expenses.category_id를 FK 후보로 설정 | 어떤 사용자와 카테고리에 연결되는지 명확히 표현하기 위해 |

# 10. 최종 개념 정리

아래 문장을 본인의 말로 완성합니다.

```text
PostgreSQL은 데이터를 관리하고 저장하는 관게형 데이터베이스 관리 시스템이다.

DBeaver 또는 psql은 sql을 쉽게 사용하게 해주는 도구이다.

데이터베이스와 스키마의 차이는 데이터베이스는 데이터를 저장하고 스키마는 그 안에서 테이블과 객체를 관리한다 .

테이블 한 행은 데이터의 속성 이다.

조회 결과가 원본 테이블과 다른 이유는 조건문을 사용하면 조건을 충족한 데이터만 보여주기 때문이다.

내부 식별자와 업무 식별자의 차이는 내부식별자는 데이터베이스 안에서 데이터를 구분하는 고유한 값이고, 업무 식별자는 다른 테이블에서 쓰이는 값이다.

PK는 데이터베이스에서 식별이 가능한 고유한 키값이다.

FK는 다른 테이블에서 스이는 연결하기 위한 키값이다.
```
