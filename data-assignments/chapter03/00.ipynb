## 3-1. 결과 기록

| 확인 항목 | 실제 결과 | 내가 이해한 의미 |
| --- | --- | --- |
| `version()` | PostgreSQL 18.4 on x86_64-windows, compiled by msvc-19.44.35227, 64-bit | 현재 SQL의 버전 확인 |
| `current_database()` | ch03 | 현재 사용중인 DB명 |
| `current_user` | postgres | 현재 사용중인 DB user명 |
| `current_schema()` | public | 현재 사용중인 스키마  |
| `search_path` | "$user", public | PostgreSQL이 테이블이나 객체를 찾을 때 어떤 스키마 순서로 검색하는지 확인하는 명령어 |
| `transaction_read_only` | off | 트랜젝션이 읽기전용인지 확인하는 명령어 |
| `TimeZone` | Asia/Seoul | DB에 입력된 지역/시간 |

## 3-2. 반드시 설명할 것

### DBeaver 연결 이름과 `current_database()`는 왜 같은 개념이 아닌가요?

```DBeaver 연결 이름은 사용자가 화면에서 구분하기 위해 임의로 정한 이름이고, current_database()는 PostgreSQL이 실제로 현재 연결되어 있는 데이터베이스 이름을 반환한다. 따라서 화면에 표시된 연결 이름만 보고 현재 데이터베이스를 판단하면 안 된다.
```

### `current_schema()`와 `search_path`는 어떤 관계가 있나요?

```search_path는 테이블 이름에 스키마를 생략했을 때 PostgreSQL이 어떤 스키마부터 찾아볼지 정한 검색 순서이다. current_schema()는 현재 search_path에서 실제로 사용 가능한 첫 번째 스키마를 반환한다.

```

### `transaction_read_only = off`라는 결과만으로 모든 테이블을 만들 권한이 있다고 단정할 수 있나요?

```단정할 수 없다. off는 현재 트랜잭션이 읽기 전용 상태가 아니라는 뜻일 뿐이며, 실제로 테이블을 생성할 수 있는지는 사용자에게 해당 데이터베이스와 스키마에 CREATE 권한이 있는지 추가로 확인해야 한다.
```

## 3-3. 증거 화면

권장 경로:

```text
assignments/chapter03/images/step03_location_check.png
```

# 5. SQL 실행 범위 실험

SQL Editor에 다음 세 문장을 입력합니다.

```sql
SELECT 'A' AS step;
SELECT 'B' AS step;
SELECT 'C' AS step;
```

## 5-1. 한 문장 실행

```text
내가 실행한 문장:
SELECT 'A' AS step;
실제 결과:A
```

## 5-2. 선택 영역 실행

```text
선택한 문장:
SELECT 'A' AS step;
SELECT 'B' AS step;
실제 결과:B
```

## 5-3. 전체 스크립트 실행

```text
실제 결과:
SELECT 'A' AS step;
SELECT 'B' AS step;
SELECT 'C' AS step;
결과 탭 또는 실행 순서에서 관찰한 점:C
```

## 5-4. 결과 해석

```text
한 문장 실행과 전체 스크립트 실행의 차이: 
전체 스크립트를 실행하면 작성된 여러 SQL 문장이 순서대로 실행되고,
각 문장의 실행 결과를 각각 확인할 수 있었다.

변경 SQL에서 실행 범위를 잘못 선택하면 위험한 이유:
INSERT, UPDATE, DELETE 같은 데이터 변경 SQL이 포함되어 있을 경우
원하지 않는 문장까지 함께 실행되어 데이터가 잘못 추가되거나 수정·삭제될 수 있기 때문이다.
```
