## 3-4. JOIN 상세 결과 관찰

```text
학생 A가 나타난 행 수: 2
학생 B가 나타난 행 수: 1
학생 C가 나타난 행 수: 0
```

다음 질문에 답합니다.

### Q1. 학생 C가 왜 결과에서 사라졌나요?

```학생 C는 질문 데이터가 없기 때문에 INNER JOIN 조건에 맞는 행이 없어 결과에서 제외되었다.

```

### Q2. 학생 A가 왜 여러 행으로 나타났나요?

```학생 A는 질문을 2개 가지고 있어서 질문 테이블의 각 행과 연결되며 2개의 행으로 나타났다.

```

### Q3. 위 `COUNT(*)`가 실제로 세고 있었던 것은 무엇인가요?

```학생 수가 아니라 학생과 질문을 JOIN한 뒤 만들어진 결과 행의 개수를 세고 있었다.

```

### Q4. 결과 숫자가 우연히 실제 학생 수와 같아도 바로 믿으면 안 되는 이유는 무엇인가요?

```우연히 숫자가 같을 뿐 SQL이 세고 있는 대상의 의미가 다를 수 있기 때문이다.
데이터가 변경되면 바로 잘못된 결과가 나타날 수 있다.

```

## 3-5. 질문 한 건을 추가한 뒤 다시 실행

### 결과 해석

```text
데이터가 바뀐 뒤 무엇이 달라졌는지: 질문이 1개 추가되면서 JOIN 결과가 3행에서 4행으로 증가했다

실제 학생 수는 왜 그대로인지: 학생 테이블에는 새로운 학생을 추가하지 않았기 때문에 학생 수는 여전히 3명이다.

이 실험을 통해 알게 된 점: SQL이 정상 실행되고 처음에는 결과 숫자가 맞아 보여도 실제로 어떤 대상을 세고 있는지 확인해야 한다.
```

# 4. 잘못된 데이터가 결과를 바꾸는 과정

## 4-1. 주문 데이터 확인

다음 데이터를 입력한 뒤 실제 결과를 확인합니다.

```text
ORD-001, 100000
ORD-002, 200000
ORD-002, 200000
```

## 4-2. 실행 전 예상

업무적으로 주문이 `ORD-001`, `ORD-002` 두 건이고 `order_code`가 주문을 고유하게 구분한다고 **가정할 경우** 예상 합계:

```text
300000 원
``` 

> 위 가정 자체가 실제 업무 규칙인지 확인해야 한다는 점도 기억합니다.

## 4-3. SQL 합계

```sql
SELECT SUM(amount) AS total_amount
FROM ch01_orders;
```

```text
500000   원
```

```아니오. SUM 함수는 저장되어 있는 세 행의 amount 값을 정확하게 더했다.

```

### 결과 차이의 원인은 무엇인가요?

```ORD-002 데이터가 두 번 저장되어 있기 때문이다. SQL은 두 행을 각각 다른 데이터로 보고 모두 합계에 포함했다.

```

### SQL이 정확하게 실행되어도 업무 숫자가 잘못될 수 있는 이유를 설명하세요.

```SQL이 정확하게 실행되어도 업무 숫자가 잘못될 수 있는 이유를 설명하세요.

```

## 4-5. AI에게 같은 데이터를 분석시킨 결과

AI가 계산한 합계:

```SQL은 데이터베이스에 저장되어 있는 데이터를 기준으로 계산하기 때문에 원본 데이터에 중복이나 오류가 있으면 SQL 문법이 정확하더라도 실제 업무에서 원하는 결과와 다른 숫자가 나올 수 있다.

```

AI가 중복 가능성을 언급했나요?

```네. ORD-002가 동일한 주문번호로 두 번 존재하기 때문에 중복 데이터일 가능성이 있다고 판단할 수 있다.
```

AI가 `ORD-002` 두 행이 실제 두 주문인지 중복 입력인지 스스로 확정할 수 있나요? 이유도 적으세요.

```아니오. 데이터만 보고는 같은 주문번호를 가진 두 행이 실제로 각각 존재해야 하는 데이터인지, 실수로 중복 입력된 것인지 확정할 수 없다. 실제 업무 규칙을 확인해야 한다.

```

추가로 확인해야 할 업무 규칙:

```
order_code는 주문마다 반드시 고유해야 하는가?
같은 order_code가 여러 번 저장될 수 있는 경우가 존재하는가?
중복 주문이 발견되었을 때 어떤 행을 정상 데이터로 판단해야 하는가?
```

내가 최종적으로 내린 판단:

```
현재 업무적으로 ORD-001과 ORD-002가 각각 하나의 주문이라고 가정한다면 예상 합계는 300000원이다. 현재 데이터베이스에는 ORD-002가 중복되어 있어 SQL 결과가 500000원으로 계산되었다. 따라서 중복 여부와 주문번호의 고유성 규칙을 먼저 확인해야 한다.

```

## 7-2. AI의 주요 제안과 나의 판단

최소 5개를 작성합니다.
1. current_database()는 현재 데이터베이스 경로가 아니라 현재 접속 중인 데이터베이스 이름이라는 설명

3. 질문이 있는 학생 수를 3명이 아니라 2명으로 수정
학생 A와 B만 질문이 있고 학생 C는 질문이 없기 때문이다.

4. INNER JOIN 결과에서 학생 C가 사라지는 이유를 설명
INNER JOIN은 양쪽 테이블에 연결되는 데이터가 있는 경우만 결과에 포함하기 때문이다.

5. COUNT(*) 결과가 학생 수가 아니라 JOIN 결과 행 수일 수 있다는 설명
실제로 질문 수가 바뀌자 결과가 3에서 4로 변경되었기 때문이다.

6. 중복 주문 데이터가 있다면 바로 삭제해야 한다는 판단
실제 중복인지 정상 데이터인지 업무 규칙을 먼저 확인해야 하기 때문이다.



## 7-3. AI에게 자기 답변을 다시 비판하도록 요청한 결과

첫 번째 답변과 달라진 점:

```
처음에는 SQL 결과 숫자만 보고 답을 작성할 수 있다고 생각했지만, 다시 검토하면서 숫자가 같더라도 SQL이 실제로 세고 있는 대상이 다를 수 있다는 점을 확인했다.

```

AI가 처음에 임의로 가정했던 내용:

```
ORD-002가 두 번 존재한다는 이유만으로 중복 데이터라고 확정할 수 있다고 볼 수 있었지만, 실제로는 같은 주문번호가 여러 행으로 존재할 수 있는 업무 규칙이 있는지 먼저 확인해야 한다.

```

추가로 업무 담당자에게 확인해야 한다고 판단한 내용:

```
order_code가 반드시 고유해야 하는지, 같은 주문번호가 여러 번 저장될 수 있는 경우가 있는지, 
중복 데이터가 발생했을 때 어떤 기준으로 정상 데이터를 판단하는지 확인해야 한다.

```

## 7-4. AI 활용에 대한 나의 평가

```text
가장 유용했던 부분: 내가 작성한 SQL과 답안에서 잘못 이해한 개념을 빠르게 찾아준 점이다.

수정하거나 거절한 부분: AI가 데이터만 보고 업무 규칙까지 확정하려는 경우에는 그대로 받아들이지 않았다.

그렇게 판단한 근거: AI는 데이터 구조와 결과는 분석할 수 있지만 실제 업무 규칙까지 항상 알고 있는 것은 아니기 때문이다.

AI가 없었다면 놓쳤을 가능성이 있는 질문: SQL 결과 숫자가 맞아 보여도 실제로 어떤 행을 세고 있는지 확인해야 한다는 질문이다.
```
